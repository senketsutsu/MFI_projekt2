# Generating data

```python
import numpy as np 
import pandas as pd 
from math import gcd
from fractions import Fraction
from mpmath import zeta
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
def save_dataset(x, y, filename):
    df = pd.DataFrame({
        "x": x,
        "y": y
    })
    df.to_csv(filename, index=False)
    print(f"Saved: {filename}")
```

```python
def save_plot(x, y, title):
    plt.figure(figsize=(10, 5))

    plt.plot(x, y)

    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")

    plt.grid(True)

    plt.show()
```

```python
x = np.linspace(-100, 100, 3000)
y = 2 * x + 1

save_dataset(x, y, "linear.csv")
save_plot(x, y, "Linear Function")
```

```python
y = x**2 - 3*x + 2

save_dataset(x, y, "quadratic.csv")
save_plot(x, y, "Quadratic Function")
```

```python
x = np.linspace(-1*np.pi, 1*np.pi, 3000)
y = np.sin(x)

save_dataset(x, y, "sinusoidal.csv")
save_plot(x, y, "Sinusoidal Function")
```

```python
x = np.linspace(-100, 100, 3000)

noise = np.random.normal(0, 8, size=len(x))

y = x**3 - 2*x**2 + x + noise

save_dataset(x, y, "noisy_polynomial.csv")
save_plot(x, y, "Noisy Polynomial")
```

```python
def weierstrass(x, a=0.5, b=3, n_terms=20):
    result = np.zeros_like(x)
    x_scaled = x /100

    for n in range(n_terms):
        result += (a**n) * np.cos((b**n) * np.pi * x_scaled) 

    return result
```

```python
x = np.linspace(-100, 100, 3000)
y = weierstrass(x)

save_dataset(x, y, "weierstrass.csv")
save_plot(x, y, "Weierstrass Function")
```

```python
n = 3000

x = np.linspace(-100, 100, n)

steps = np.random.normal(0, 0.03, size=n)

y = np.cumsum(steps)

save_dataset(x, y, "wiener.csv")
save_plot(x, y, "Wiener Process")
```

```python
x = np.linspace(1.1, 100, 3000)

y = np.array([float(zeta(val)) for val in x])

save_dataset(x, y, "riemann_zeta.csv")
save_plot(x, y, "Riemann Zeta Function")
```
